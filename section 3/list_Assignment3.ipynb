```python
#q10
n=int(input("enter number of sublists"))
nest=[list(map(int,input().split()))for i in range(n)]
print(nest)
flat=[]
for i in nest:
    for j in i:
        flat.append(j)
print(flat)
```

```python
#q14 
def revn(og, n):
    print(og)
    for i in range(n):
        a=og.pop()
        og.insert(0, a)
    print(og)

og=[1,2,3,4]
revn(og, 2)
```

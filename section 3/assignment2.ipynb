```python
#q15
n = int(input("How many Fibonacci terms? "))

a = 0  
b = 1 
count = 0  # how many numbers printed so far

while count < n:
    print(a, end=" ")  # print the current number
    c = a + b          # next number is sum of a and b
    a = b              # move a to the next one
    b = c              # move b to the next one3
    count += 1         # update how many numbers we’ve done
```
